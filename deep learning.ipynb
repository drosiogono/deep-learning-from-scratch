## Gradient Descent Practice

```python
## gradient descent practice
# 함수 f(x) 정의
def f(x):
    return x**2 - 4*x + 5

# f(x)의 도함수 정의
def df(x):
    return 2*x - 4

# 경사 하강법 알고리즘 구현
def gradient_descent(initial_x, learning_rate, num_iterations):
    x = initial_x
    x_history = [x]  # x 값의 변화를 추적하기 위한 리스트
    y_history = [f(x)]
    for i in range(num_iterations):
        gradient = df(x)
        x = x - learning_rate * gradient
        x_history.append(x)
        y_history.append(f(x))
    return x, x_history, y_history

# 하이퍼파라미터 설정
initial_x = 0  # 초기 x값
learning_rate = 0.1  # 학습률
num_iterations = 100  # 반복 횟수

# 경사 하강법 실행
x_min, x_history, y_history = gradient_descent(initial_x, learning_rate, num_iterations)

# 결과 출력
print(f"최솟값: {x_min:.4f}")
print(f"함수 최솟값: {f(x_min):.4f}")

# x 값의 변화 시각화
plt.figure(figsize=(10, 6))
plt.plot(x_history, y_history,'-o')

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Gradient Descent')
plt.show()
```

```python
def normalize(a, mean=None, std=None):
  if mean is None:
    mean = a.mean(axis=0)
  if std is None:
    std = a.std(axis=0)
  std += np.finfo(float).eps
  return (a - mean) / std, mean, std
```

## linear regression

```python
class Unit:
    def __init__(self):
        self.w = None
        self.b = None
        self.log = []
    def fit(self, x, y, epoch=10000, lr=0.01):
        x = normalize(x.to_numpy())
        y = y.to_numpy().reshape(-1, 1) # Reshape the NumPy array
        self.w = np.ones((np.shape(x)[1], 1))
        self.b = 1
        for _ in range(0, epoch):
            y_hat = x @ self.w + self.b # forward
            err = -(y - y_hat)
            self.w -= lr * (x.T @ err) / len(x) # backpropagation
            self.b -= lr * np.sum(err) / len(x) # backpropagation
            self.log.append(np.mean(err))
```

## 1-layer Logistic Regression

```python
def sigmoid(z):
  z = np.clip(z, -500, 500)
  return 1 / (1 + np.exp(-z))

class Logistic:
  def __init__(self):
    self.w = None
    self.b = None
    self.mean = None
    self.std = None
  def fit(self, x, y, lr=0.01, epoch=1000):
    x, self.mean, self.std = normalize(x)
    self.w = np.zeros((np.shape(x)[1], 1))
    y = y.reshape(-1, 1)
    self.b = 0
    for i in range(epoch):
      z = x @ self.w + self.b # forward
      y_hat = sigmoid(z) # final activation?
      err = y_hat - y
      self.w -= lr * (x.T @ err) / len(x) # backpropagation
      self.b -= lr * np.mean(err)
```

## 2-layer Logistic Regression

```python
def relu(n):
  return np.maximum(0, n)

def drelu(n):
  return (n > 0).astype(float)

def sigmoid(z):
  # z = np.clip(z, -500, 500)
  return 1 / (1 + np.exp(-z))
```

```python
class DLL:
  def __init__(self):
    self.w = None
    self.b = None
    self.mean = None
    self.std = None
  def fit(self, x, y, nhidden=5, lr=0.01, epoch=1000):
    eps = np.finfo(float).eps
    x, self.mean, self.std = normalize(x)
    y = y.reshape(-1, 1)
    self.w1 = np.random.randn(np.shape(x)[1], nhidden) * 0.01
    self.b1 = np.zeros(nhidden)
    self.w2 = np.random.randn(nhidden, 1) * 0.01
    self.b2 = 0
    self.bce = []
    for _ in range(epoch):
      z1 = x @ self.w1 + self.b1
      a1 = relu(z1)
      z2 = a1 @ self.w2 + self.b2
      y_hat = sigmoid(z2)
      self.bce.append(-np.mean(y * np.log(y_hat + eps) + (1 - y) * np.log(1 - y_hat + eps)))

      d2 = y_hat - y
      d1 = (d2 @ self.w2.T) * drelu(z1)
      self.b2 -= lr * np.mean(d2)
      self.w2 -= lr * (a1.T @ d2) / len(a1)
      self.b1 -= lr * np.mean(d1, axis=0)
      self.w1 -= lr * (x.T @ d1) / len(x)
```

```python
def relu(n):
  return np.maximum(0, n)

def drelu(n):
  return (n > 0).astype(float)

def sigmoid(z):
  z = np.clip(z, -500, 500)
  return 1 / (1 + np.exp(-z))

def bce(y, y_hat, w0=None, w1=None):
  eps = np.finfo(float).eps
  if w0 is None:
    w0 = y.shape[0] / (2 * np.sum(y == 0))
  if w1 is None:
    w1 = y.shape[0] / (2 * np.sum(y == 1))
  return -np.mean(w1 * y * np.log(y_hat + eps) + w0 * (1 - y) * np.log(1 - y_hat + eps))
  # 처음에는 w0, w1을 안 곱하고 했지만 제대로 학습이 되지 않았음. 모델이 하나로만 출력
  # 가중치가 있는 BCE가 필요하다고 생각해서 구현함
```

weighted BCE의 나만의 해석


(y.shape[0] / 2) / np.sum(y = x) 꼴

비율을 맞추려면 np.sum(y = x)로 나누는 게 직관적임

대신 차원을 맞추려면? 1 / np.sum(y = x)에 뭔가 곱해야 함

이상적인 상황에서 np.sum(y = x) = y.shape[0] / 2니까

(y.shape[0] / 2) / np.sum(y = x)가 됨

```python
class DLL:
  def __init__(self):
    self.w = None
    self.b = None
    self.mean = None
    self.std = None

  def fit(self, x, y, nhidden=5, lr=0.01, epoch=1000):
    eps = np.finfo(float).eps
    x, self.mean, self.std = normalize(x)
    y = y.reshape(-1, 1)

    self.w1 = np.random.randn(np.shape(x)[1], nhidden) * np.sqrt(2 / np.shape(x)[1])
    self.b1 = np.zeros(nhidden)
    self.w2 = np.random.randn(nhidden, 1) * np.sqrt(2 / nhidden)
    self.b2 = 0
    self.bce = []

    w0 = y.shape[0] / (2 * np.sum(y == 0))   # ≈ 2.51
    w1 = y.shape[0] / (2 * np.sum(y == 1))  # ≈ 0.62

    for _ in range(epoch):
      z1, a1, z2, y_hat = self.forward(x)
      self.bce.append(bce(y, y_hat))

      d2 = (y_hat - y) * ((w1 * y + w0 * (1 - y)))
      d1 = (d2 @ self.w2.T) * drelu(z1)
      self.b2 -= lr * np.mean(d2)
      self.w2 -= lr * (a1.T @ d2) / len(a1)
      self.b1 -= lr * np.mean(d1, axis=0)
      self.w1 -= lr * (x.T @ d1) / len(x)

  def forward(self, x):
    z1 = x @ self.w1 + self.b1
    a1 = relu(z1)
    z2 = a1 @ self.w2 + self.b2
    y_hat = sigmoid(z2)
    return z1, a1, z2, y_hat

  def predict_proba(self, x):
    x = normalize(x, self.mean, self.std)[0]
    return self.forward(x)[-1]

  def predict(self, x):
    probs = self.predict_proba(x)
    return (probs >= 0.5).astype(int)
```

```python
def bce(y, y_hat, w0=None, w1=None):
  eps = np.finfo(float).eps
  if w0 is None:
    w0 = y.shape[0] / (2 * np.sum(y == 0))
  if w1 is None:
    w1 = y.shape[0] / (2 * np.sum(y == 1))
  return -np.mean(w1 * y * np.log(y_hat + eps) + w0 * (1 - y) * np.log(1 - y_hat + eps))
```

```python
class Logistic:
  def __init__(self):
    self.w = None
    self.b = None
    self.mean = None
    self.std = None
    self.bce = []
    self.bce_w = []
  def fit(self, x, y, lr=0.01, epoch=1000):
    x, self.mean, self.std = normalize(x)
    self.w = np.zeros((np.shape(x)[1], 1))
    y = y.reshape(-1, 1)
    self.b = 0
    w0 = y.shape[0] / (2 * np.sum(y == 0))   # ≈ 2.51
    w1 = y.shape[0] / (2 * np.sum(y == 1))  # ≈ 0.62
    self.bce_w = [w0, w1]

    for i in range(epoch):
      z = x @ self.w + self.b # forward
      y_hat = sigmoid(z) # final activation?
      self.bce.append(bce(y, y_hat, w0, w1))
      err = (y_hat - y) * ((w1 * y + w0 * (1 - y)))
      self.w -= lr * (x.T @ err) / len(x) # backpropagation
      self.b -= lr * np.mean(err)

  def predict_proba(self, x):
    x = normalize(x, self.mean, self.std)[0]
    return sigmoid(x @ self.w + self.b)

  def predict(self, x):
    probs = self.predict_proba(x)
    return (probs >= 0.5).astype(int)
```

## Multi-layer softmax

```python
def relu(n):
  return np.maximum(0, n)

def drelu(n):
  return (n > 0).astype(float)

def softmax(z):
  # z = np.clip(z, -500, 500)
  z = z - np.max(z, axis=1, keepdims=True)
  ez = np.exp(z)
  return ez / np.sum(ez, axis=1, keepdims=True)

def ce(y, y_hat):
  eps = np.finfo(float).eps
  # w0 = y.shape[0] / (2 * np.sum(y == 0))
  # w1 = y.shape[0] / (2 * np.sum(y == 1))
  y_hat = np.clip(y_hat, eps, 1 - eps)
  return -np.mean(np.sum(y * np.log(y_hat + eps), axis=1))

def one_hot(y):
  nres = len(np.unique(y))
  return np.eye(nres)[y]
```

```python
class Unit:
  def __init__(self, actv = relu, dactv = drelu):
    self.w = []
    self.b = []
    self.mean = None
    self.std = None
    self.ce = []
    self.activationf = softmax
    self.actv = actv
    self.dactv = dactv

  def fit(self, x, y, nhiddens: list[int], lr=0.01, epoch=1000):
    x, self.mean, self.std = normalize(x)
    nres = len(np.unique(y))
    y = one_hot(y)

    sizes = [np.shape(x)[1]] + nhiddens + [nres]
    for i in range(len(sizes) - 1):
      wi = np.random.randn(
          sizes[i], sizes[i+1]
      ) * np.sqrt(2 / sizes[i])
      bi = np.zeros(sizes[i+1])
      self.w.append(wi)
      self.b.append(bi)

    for i in range(epoch):
      activations, zs = self.forward(x)
      y_hat = activations[-1]
      self.ce.append(ce(y, y_hat))
      self.backprop(lr, activations, zs, y)

  def forward(self, x):
    activations = [x]
    zs = []
    a = x
    for i in range(len(self.w) - 1):
      z = a @ self.w[i] + self.b[i]
      a = self.actv(z)
      zs.append(z)
      activations.append(a)
    lastz = a @ self.w[-1] + self.b[-1]
    zs.append(lastz)
    y_hat = self.activationf(lastz)
    activations.append(y_hat)
    return activations, zs

  def backprop(self, lr, activations, zs, y):
    N = len(y)

    grad_w = [None for _ in range(len(self.w))]
    grad_b = [None for _ in range(len(self.b))]

    delta = activations[-1] - y
    dw = (activations[-2].T @ delta) / N
    db = np.mean(delta, axis=0)
    grad_w[-1] = dw
    grad_b[-1] = db

    for i in range(len(self.w)-2, -1, -1):
      delta = delta @ self.w[i+1].T * self.dactv(zs[i])

      dw = (activations[i].T @ delta) / N
      db = np.mean(delta, axis=0)
      grad_w[i] = dw
      grad_b[i] = db

    for i in range(len(self.w)):
      self.w[i] -= lr * grad_w[i]
      self.b[i] -= lr * grad_b[i]

  def predict_proba(self, x):
    x, _, _ = normalize(x, self.mean, self.std)
    activations, _ = self.forward(x)
    return activations[-1]

  def predict(self, x):
    preds = self.predict_proba(x)
    return np.argmax(preds, axis=1)
```

## Multi-layer softmax with Momentum, RMSProp, and Adam

### Momentum

```python
class Momentum:
  def __init__(self, beta = 0.9):
    self.beta = beta
    self.v = None
    self.t = 1

  def update(self, grad):
    if self.v is None:
      self.v = np.zeros_like(grad)
    self.v = self.beta * self.v + (1 - self.beta) * grad
    v_hat = self.v / (1 - self.beta ** self.t)
    self.t += 1
    return v_hat
```

```python
def relu(n):
  return np.maximum(0, n)

def drelu(n):
  return (n > 0).astype(float)

def softmax(z):
  # z = np.clip(z, -500, 500)
  z = z - np.max(z, axis=1, keepdims=True)
  ez = np.exp(z)
  return ez / np.sum(ez, axis=1, keepdims=True)

def ce(y, y_hat):
  eps = np.finfo(float).eps
  # w0 = y.shape[0] / (2 * np.sum(y == 0))
  # w1 = y.shape[0] / (2 * np.sum(y == 1))
  y_hat = np.clip(y_hat, eps, 1 - eps)
  return -np.mean(np.sum(y * np.log(y_hat + eps), axis=1))

def one_hot(y):
  nres = len(np.unique(y))
  return np.eye(nres)[y]
```

```python
class MUnit:
  def __init__(self, actv = relu, dactv = drelu):
    self.w = []
    self.b = []
    self.mean = None
    self.std = None
    self.ce = []
    self.activationf = softmax
    self.actv = actv
    self.dactv = dactv
    self.mw_optimizers = [] # List to hold Momentum optimizers for weights
    self.mb_optimizers = [] # List to hold Momentum optimizers for biases

  def fit(self, x, y, nhiddens: list[int], lr=0.01, epoch=1000):
    x, self.mean, self.std = normalize(x)
    nres = len(np.unique(y))
    y = one_hot(y)

    # Clear existing weights/biases/optimizers if fit is called multiple times
    self.w = []
    self.b = []
    self.mw_optimizers = []
    self.mb_optimizers = []
    self.ce = [] # Clear cross-entropy log

    sizes = [np.shape(x)[1]] + nhiddens + [nres]
    for i in range(len(sizes) - 1):
      wi = np.random.randn(
          sizes[i], sizes[i+1]
      ) * np.sqrt(2 / sizes[i])
      bi = np.zeros(sizes[i+1])
      self.w.append(wi)
      self.b.append(bi)
      self.mw_optimizers.append(Momentum()) # One optimizer per weight matrix
      self.mb_optimizers.append(Momentum()) # One optimizer per bias vector

    for i in range(epoch):
      activations, zs = self.forward(x)
      y_hat = activations[-1]
      self.ce.append(ce(y, y_hat))
      self.backprop(lr, activations, zs, y)

  def forward(self, x):
    activations = [x]
    zs = []
    a = x
    for i in range(len(self.w) - 1):
      z = a @ self.w[i] + self.b[i]
      a = self.actv(z)
      zs.append(z)
      activations.append(a)
    lastz = a @ self.w[-1] + self.b[-1]
    zs.append(lastz)
    y_hat = self.activationf(lastz)
    activations.append(y_hat)
    return activations, zs

  def backprop(self, lr, activations, zs, y):
    N = len(y)

    grad_w = [None for _ in range(len(self.w))]
    grad_b = [None for _ in range(len(self.b))]

    delta = activations[-1] - y
    dw = (activations[-2].T @ delta) / N
    db = np.mean(delta, axis=0)
    grad_w[-1] = dw
    grad_b[-1] = db

    for i in range(len(self.w)-2, -1, -1):
      delta = delta @ self.w[i+1].T * self.dactv(zs[i])

      dw = (activations[i].T @ delta) / N
      db = np.mean(delta, axis=0)
      grad_w[i] = dw
      grad_b[i] = db

    for i in range(len(self.w)):
      # Use the specific momentum optimizer for each weight and bias
      self.w[i] -= lr * self.mw_optimizers[i].update(grad_w[i])
      self.b[i] -= lr * self.mb_optimizers[i].update(grad_b[i])

  def predict_proba(self, x):
    x, _, _ = normalize(x, self.mean, self.std)
    activations, _ = self.forward(x)
    return activations[-1]

  def predict(self, x):
    preds = self.predict_proba(x)
    return np.argmax(preds, axis=1)
```

### RMSProp

```python
class RMSProp:
  def __init__(self, beta = 0.99):
    self.beta = beta
    self.s = None
    self.t = 1

  def update(self, grad):
    if self.s is None:
      self.s = np.zeros_like(grad)
    self.s = self.beta * self.s + (1 - self.beta) * (grad * grad)
    # s_hat = self.s / (1 - self.beta ** self.t)
    self.t += 1
    return grad / (np.sqrt(self.s) + np.finfo(float).eps)
```

```python
def relu(n):
  return np.maximum(0, n)

def drelu(n):
  return (n > 0).astype(float)

def softmax(z):
  # z = np.clip(z, -500, 500)
  z = z - np.max(z, axis=1, keepdims=True)
  ez = np.exp(z)
  return ez / np.sum(ez, axis=1, keepdims=True)

def ce(y, y_hat):
  eps = np.finfo(float).eps
  # w0 = y.shape[0] / (2 * np.sum(y == 0))
  # w1 = y.shape[0] / (2 * np.sum(y == 1))
  y_hat = np.clip(y_hat, eps, 1 - eps)
  return -np.mean(np.sum(y * np.log(y_hat + eps), axis=1))

def one_hot(y):
  nres = len(np.unique(y))
  return np.eye(nres)[y]
```

```python
class RUnit:
  def __init__(self, actv = relu, dactv = drelu):
    self.w = []
    self.b = []
    self.mean = None
    self.std = None
    self.ce = []
    self.activationf = softmax
    self.actv = actv
    self.dactv = dactv
    self.mw_optimizers = [] # List to hold RMSProp optimizers for weights
    self.mb_optimizers = [] # List to hold RMSProp optimizers for biases

  def fit(self, x, y, nhiddens: list[int], lr=0.01, epoch=1000):
    x, self.mean, self.std = normalize(x)
    nres = len(np.unique(y))
    y = one_hot(y)

    # Clear existing weights/biases/optimizers if fit is called multiple times
    self.w = []
    self.b = []
    self.mw_optimizers = []
    self.mb_optimizers = []
    self.ce = [] # Clear cross-entropy log

    sizes = [np.shape(x)[1]] + nhiddens + [nres]
    for i in range(len(sizes) - 1):
      wi = np.random.randn(
          sizes[i], sizes[i+1]
      ) * np.sqrt(2 / sizes[i])
      bi = np.zeros(sizes[i+1])
      self.w.append(wi)
      self.b.append(bi)
      self.mw_optimizers.append(RMSProp()) # One optimizer per weight matrix
      self.mb_optimizers.append(RMSProp()) # One optimizer per bias vector

    for i in range(epoch):
      activations, zs = self.forward(x)
      y_hat = activations[-1]
      self.ce.append(ce(y, y_hat))
      self.backprop(lr, activations, zs, y)

  def forward(self, x):
    activations = [x]
    zs = []
    a = x
    for i in range(len(self.w) - 1):
      z = a @ self.w[i] + self.b[i]
      a = self.actv(z)
      zs.append(z)
      activations.append(a)
    lastz = a @ self.w[-1] + self.b[-1]
    zs.append(lastz)
    y_hat = self.activationf(lastz)
    activations.append(y_hat)
    return activations, zs

  def backprop(self, lr, activations, zs, y):
    N = len(y)

    grad_w = [None for _ in range(len(self.w))]
    grad_b = [None for _ in range(len(self.b))]

    delta = activations[-1] - y
    dw = (activations[-2].T @ delta) / N
    db = np.mean(delta, axis=0)
    grad_w[-1] = dw
    grad_b[-1] = db

    for i in range(len(self.w)-2, -1, -1):
      delta = delta @ self.w[i+1].T * self.dactv(zs[i])

      dw = (activations[i].T @ delta) / N
      db = np.mean(delta, axis=0)
      grad_w[i] = dw
      grad_b[i] = db

    for i in range(len(self.w)):
      # Use the specific RMSProp optimizer for each weight and bias
      self.w[i] -= lr * self.mw_optimizers[i].update(grad_w[i])
      self.b[i] -= lr * self.mb_optimizers[i].update(grad_b[i])

  def predict_proba(self, x):
    x, _, _ = normalize(x, self.mean, self.std)
    activations, _ = self.forward(x)
    return activations[-1]

  def predict(self, x):
    preds = self.predict_proba(x)
    return np.argmax(preds, axis=1)
```

### Adam

```python
class Adam:
  def __init__(self, beta1 = 0.9, beta2 = 0.99):
    self.beta1 = beta1
    self.beta2 = beta2
    self.v = None
    self.s = None
    self.t = 1

  def update(self, grad):
    if self.s is None:
      self.s = np.zeros_like(grad)
    if self.v is None:
      self.v = np.zeros_like(grad)
    self.v = self.beta1 * self.v + (1 - self.beta1) * grad
    self.s = self.beta2 * self.s + (1 - self.beta2) * (grad * grad)

    v_hat = self.v / (1 - self.beta1 ** self.t)
    s_hat = self.s / (1 - self.beta2 ** self.t)

    self.t += 1
    return v_hat / (np.sqrt(s_hat) + np.finfo(float).eps)
```

## Multi-layer softmax with Dropout

```python
class Dropout:
  def __init__(self, p = 0.5):
    self.p = p
    self.mask = None

  def forward(self, x, is_training = True):
    if is_training:
      self.mask = np.random.rand(*x.shape) > self.p
      return x * self.mask / (1.0 - self.p)

  def backward(self, delta):
    return delta * self.mask
```

```python
def relu(n):
  return np.maximum(0, n)

def drelu(n):
  return (n > 0).astype(float)

def softmax(z):
  # z = np.clip(z, -500, 500)
  z = z - np.max(z, axis=1, keepdims=True)
  ez = np.exp(z)
  return ez / np.sum(ez, axis=1, keepdims=True)

def ce(y, y_hat):
  eps = np.finfo(float).eps
  # w0 = y.shape[0] / (2 * np.sum(y == 0))
  # w1 = y.shape[0] / (2 * np.sum(y == 1))
  y_hat = np.clip(y_hat, eps, 1 - eps)
  return -np.mean(np.sum(y * np.log(y_hat + eps), axis=1))

def one_hot(y):
  nres = len(np.unique(y))
  return np.eye(nres)[y]
```

```python
class Unit:
  def __init__(self, actv = relu, dactv = drelu):
    self.w = []
    self.b = []
    self.mean = None
    self.std = None
    self.ce = []
    self.activate = softmax
    self.actv = actv
    self.dactv = dactv

  def fit(self, x, y, nhiddens: list[int], lr=0.01, epoch=1000, p = 0.3):
    x, self.mean, self.std = normalize(x)
    nres = len(np.unique(y))
    y = one_hot(y)
    self.dropout = Dropout(p)

    sizes = [np.shape(x)[1]] + nhiddens + [nres]
    for i in range(len(sizes) - 1):
      wi = np.random.randn(
          sizes[i], sizes[i+1]
      ) * np.sqrt(2 / sizes[i])
      bi = np.zeros(sizes[i+1])
      self.w.append(wi)
      self.b.append(bi)

    for i in range(epoch):
      activations, zs = self.forward(x)
      y_hat = activations[-1]
      self.ce.append(ce(y, y_hat))
      self.backprop(lr, activations, zs, y)

  def forward(self, x):
    activations = [x]
    zs = []
    a = x
    for i in range(len(self.w) - 1):
      h = a @ self.w[i] + self.b[i]
      z = self.dropout.forward(h)
      a = self.actv(z)
      zs.append(z)
      activations.append(a)
    lastz = a @ self.w[-1] + self.b[-1]
    zs.append(lastz)
    y_hat = self.activate(lastz)
    activations.append(y_hat)
    return activations, zs

  def backprop(self, lr, activations, zs, y):
    N = len(y)

    grad_w = [None for _ in range(len(self.w))]
    grad_b = [None for _ in range(len(self.b))]

    delta = activations[-1] - y
    dw = (activations[-2].T @ delta) / N
    db = np.mean(delta, axis=0)
    dw = self.dropout.backward(dw)
    db = self.dropout.backward(db)
    grad_w[-1] = dw
    grad_b[-1] = db

    for i in range(len(self.w)-2, -1, -1):
      delta = delta @ self.w[i+1].T * self.dactv(zs[i])

      dw = (activations[i].T @ delta) / N
      db = np.mean(delta, axis=0)
      dw = self.dropout.backward(dw)
      db = self.dropout.backward(db)
      grad_w[i] = dw
      grad_b[i] = db

    for i in range(len(self.w)):
      self.w[i] -= lr * grad_w[i]
      self.b[i] -= lr * grad_b[i]

  def predict_proba(self, x):
    x, _, _ = normalize(x, self.mean, self.std)
    activations, _ = self.forward(x)
    return activations[-1]

  def predict(self, x):
    preds = self.predict_proba(x)
    return np.argmax(preds, axis=1)
```

## Multi-layer softmax with weight decay

```python
def relu(n):
  return np.maximum(0, n)

def drelu(n):
  return (n > 0).astype(float)

def softmax(z):
  # z = np.clip(z, -500, 500)
  z = z - np.max(z, axis=1, keepdims=True)
  ez = np.exp(z)
  return ez / np.sum(ez, axis=1, keepdims=True)

def ce(y, y_hat):
  eps = np.finfo(float).eps
  # w0 = y.shape[0] / (2 * np.sum(y == 0))
  # w1 = y.shape[0] / (2 * np.sum(y == 1))
  y_hat = np.clip(y_hat, eps, 1 - eps)
  return -np.mean(np.sum(y * np.log(y_hat + eps), axis=1))

def one_hot(y):
  nres = len(np.unique(y))
  return np.eye(nres)[y]
```

```python
class WDUnit:
  def __init__(self, actv = relu, dactv = drelu, lambd = 0.0001):
    self.w = []
    self.b = []
    self.mean = None
    self.std = None
    self.ce = []
    self.activationf = softmax
    self.actv = actv
    self.dactv = dactv
    self.lambd = lambd

  def fit(self, x, y, nhiddens: list[int], lr=0.01, epoch=1000):
    x, self.mean, self.std = normalize(x)
    nres = len(np.unique(y))
    y = one_hot(y)

    sizes = [np.shape(x)[1]] + nhiddens + [nres]
    for i in range(len(sizes) - 1):
      wi = np.random.randn(
          sizes[i], sizes[i+1]
      ) * np.sqrt(2 / sizes[i])
      bi = np.zeros(sizes[i+1])
      self.w.append(wi)
      self.b.append(bi)

    for i in range(epoch):
      activations, zs = self.forward(x)
      y_hat = activations[-1]
      self.ce.append(ce(y, y_hat))
      self.backprop(lr, activations, zs, y)

  def forward(self, x):
    activations = [x]
    zs = []
    a = x
    for i in range(len(self.w) - 1):
      z = a @ self.w[i] + self.b[i]
      a = self.actv(z)
      zs.append(z)
      activations.append(a)
    lastz = a @ self.w[-1] + self.b[-1]
    zs.append(lastz)
    y_hat = self.activationf(lastz)
    activations.append(y_hat)
    return activations, zs

  def backprop(self, lr, activations, zs, y):
    N = len(y)

    grad_w = [None for _ in range(len(self.w))]
    grad_b = [None for _ in range(len(self.b))]

    delta = activations[-1] - y
    dw = (activations[-2].T @ delta) / N
    db = np.mean(delta, axis=0)
    grad_w[-1] = dw
    grad_b[-1] = db

    for i in range(len(self.w)-2, -1, -1):
      delta = delta @ self.w[i+1].T * self.dactv(zs[i])

      dw = (activations[i].T @ delta) / N
      db = np.sum(delta, axis=0) / N
      grad_w[i] = dw
      grad_b[i] = db

    for i in range(len(self.w)):
      self.w[i] = (1 - self.lambd * lr / N) * self.w[i] - lr * grad_w[i]
      self.b[i] -= lr * grad_b[i]

  def predict_proba(self, x):
    x, _, _ = normalize(x, self.mean, self.std)
    activations, _ = self.forward(x)
    return activations[-1]

  def predict(self, x):
    preds = self.predict_proba(x)
    return np.argmax(preds, axis=1)
```
